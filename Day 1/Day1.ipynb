### 🟢 Problem: Find the Second Largest Element in an Array

Example:

arr = [12, 45, 7, 89, 34]

Expected Output:
Second Largest = 45

```python
# Find a Largest Number

def largest(arr):
    
    if len(arr)<=0:
        largest=arr[0]
        for i in len(arr):

            if arr[i]>largest:
                largest=arr[i]

    largest(arr)        

arr=[12,45,7,89,34]

```

```python
# Find a Second Largest

def second_largest(arr):

    # Check if the array is empty
    if len(arr) < 2:
        return None

    # Initially assume first element is largest
    largest_number = arr[0]
    second_number = None

    # Traverse the array
    for i in range(1, len(arr)):

        # If current number is greater than largest
        if arr[i] > largest_number:

            # Old largest becomes second largest
            second_number = largest_number

            # Current number becomes largest
            largest_number = arr[i]

        # If current number can become second largest
        elif arr[i] < largest_number:

            if second_number is None or arr[i] > second_number:
                second_number = arr[i]

    return second_number


arr = [12, 45, 7, 89, 34]

print("Second Largest Number:", second_largest(arr))
```
