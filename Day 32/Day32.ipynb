## 🔥 Day 32 — Rotate an Array to the Right by K Positions
Given an array and an integer k, rotate the array to the right by k positions.
Example
arr = [1, 2, 3, 4, 5]

k = 2

After rotating right by 2:

[4, 5, 1, 2, 3]

Another example

arr = [10, 20, 30, 40, 50, 60]

k = 3

Output:

[40, 50, 60, 10, 20, 30]

```python
def rotate(arr):
    k=2
    last=0
    for i in range(k):
        last=arr[-1]
        last=arr.pop()
        arr.insert(0,last)

    return arr
arr=[1,2,3,4,5]
print("The Rotated Array is:",rotate(arr))
```
