# 🔥 Day 52: Divide Two Integers

**Problem Statement:**

Given two integers `dividend` and `divisor`, divide the two integers without using multiplication (`*`), division (`/`), or modulo (`%`) operators.

The integer division should truncate toward zero, meaning the fractional part is discarded.

For example:
- `8.345` → `8`
- `-2.7335` → `-2`

Return the quotient after dividing `dividend` by `divisor`.

**Constraints:**
- Assume the environment supports only 32-bit signed integers: `[-2³¹, 2³¹ − 1]`.
- If the quotient exceeds `2³¹ − 1`, return `2³¹ − 1`.
- If the quotient is less than `−2³¹`, return `−2³¹`.
- Do not use multiplication, division, or modulo operators.

**Example 1:**

Input:
```python
dividend = 10
divisor = 3
```

Output:
```python
3
```

Explanation: `10 / 3 = 3.333...`, which truncates to `3`.

**Example 2:**

Input:
```python
dividend = 7
divisor = -3
```

Output:
```python
-2
```

Explanation: `7 / -3 = -2.333...`, which truncates toward zero to `-2`.




````python
def divide(dividend, divisor):
    result=0
    if divisor==0:
        print("Error")
        return

    while dividend>=divisor:
        dividend=dividend-divisor
        result+=1

    return result


dividend=1000000000
divisor=1
print("The Result is:",divide(dividend,divisor))
````

````python
def divide(dividend, divisor):
    result=0
    shift=0
    negative = (dividend < 0) ^ (divisor < 0)
    dividend = abs(dividend)
    divisor = abs(divisor)
    
    if divisor==0:
        print("Error")
        return

    while (divisor<<shift)<=dividend:
        dividend=dividend-(divisor<<shift)
        result=result+(1<<shift)

    if negative:
            result = -result
    return result


dividend=7
divisor=-3
print("The Result is:",divide(dividend,divisor))
````
