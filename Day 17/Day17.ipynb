# 🧩 Day 17 — Find the Pair with the Maximum Sum

## Given an array, find the two different elements whose sum is maximum.

Example:

Input:

[5, 2, 9, 1, 7]

Output:

16

Because:

9 + 7 = 16

```python
def maximum_pair_sum(arr):

    if len(arr) < 2:
        return None

    largest_number = arr[0]
    second_number = None

    for i in range(1, len(arr)):
        if arr[i] > largest_number:
            second_number = largest_number
            largest_number = arr[i]
        elif arr[i] < largest_number:
            if second_number is None or arr[i] > second_number:
                second_number = arr[i]

    return largest_number+second_number


arr = [12, 45, 7, 89, 34]

print("The Maximum Sum:", maximum_pair_sum(arr))
```
