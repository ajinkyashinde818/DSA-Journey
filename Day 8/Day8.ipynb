# 🧩 Day 8 — First Non-Repeating Element

## Given an array, find the first element that appears only once.

Example 1
arr = [4, 5, 1, 2, 1, 4]

Answer → 5

Because:

4 → repeated ❌
5 → appears once ✅ → first non-repeating

```python
def non_repeat(arr):
     for i in range(len(arr)):
        count=0
        for j in range(len(arr)):
            if arr[i]==arr[j]:
                count=count+1
            
        if count==1:
            return arr[i]
            
arr=[4,5,1,2,1,4]  
print("The Non Repeatable Number is:",non_repeat(arr))  
```
