### 🔥 Day 46 — 3Sum (Find Triplets That Sum to Zero)

Given an integer array `nums`, return all unique triplets `[nums[i], nums[j], nums[k]]` such that:

- `i != j`, `i != k`, and `j != k`
- `nums[i] + nums[j] + nums[k] == 0`
- The result must not contain duplicate triplets.

**Example 1:**

```python
Input: nums = [-1, 0, 1, 2, -1, -4]
Output: [[-1, -1, 2], [-1, 0, 1]]
```

**Explanation:** Both `[-1, -1, 2]` and `[-1, 0, 1]` have a sum of zero. Duplicate triplets must be excluded.

**Example 2:**

```python
Input: nums = [0, 1, 1]
Output: []
```

**Explanation:** No three elements sum to zero.

**Example 3:**

```python
Input: nums = [0, 0, 0]
Output: [[0, 0, 0]]
```

**Explanation:** All three elements sum to zero, producing one unique triplet.

**Constraints:**
- `3 <= len(nums) <= 3000`
- `-100000 <= nums[i] <= 100000`

**Your Task:** Find all unique triplets whose sum is zero, without including duplicate triplets.

````python
def three_sum(num):
    result=[]
    num.sort()

    for i in range(len(num)):
        if i > 0 and num[i] == num[i - 1]:
            continue

        left=i+1
        right=len(num)-1

        while left<right:
            
            sum = num[i]+num[left]+num[right]

            if sum==0:
                result.append([num[i], num[left], num[right]])
                left += 1
                right -= 1
            elif sum<0:
                left+=1
            else:
                right-=1

    return result

num=[-1,0,1,2,-1,-4]
print("The Numbers is:",three_sum(num))
````
