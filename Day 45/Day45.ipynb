### 🔥 Day 45 — Palindrome Number

Given an integer `x`, return `True` if `x` is a **palindrome**, and `False` otherwise.

A palindrome number reads the same from left to right and from right to left.

**Examples:**

**Example 1:**
```python
Input: x = 121
Output: True
```
**Explanation:** `121` reads the same in both directions.

**Example 2:**
```python
Input: x = -121
Output: False
```
**Explanation:** From left to right, it reads `-121`. From right to left, it becomes `121-`, so it is not a palindrome.

**Example 3:**
```python
Input: x = 10
Output: False
```
**Explanation:** Reversing the digits gives `01`, which is not equal to `10`.

**Example 4:**
```python
Input: x = 0
Output: True
```
**Explanation:** A single-digit number is always a palindrome.

**Constraints:**
- `-2³¹ <= x <= 2³¹ - 1`
- Negative numbers are not palindromes.

**Your Task:** Determine whether the given integer is a palindrome without converting it into a string."

````python
def palindrome(num):
    if num<0:
            return False
    
    original=num
    rev=0
    while num>0:
        last=num%10
        rev=(rev*10)+last
        num=num//10

    if rev==original:
        return True
    else:
        return False
    
num=121
print("The Result is:",palindrome(num))
````
