🎯 Merge Two Sorted Arrays

Given two sorted arrays, merge them into one sorted array.

Example:

arr1 = [1, 3, 5, 7]

arr2 = [2, 4, 6, 8]

Output:

[1, 2, 3, 4, 5, 6, 7, 8]

```python
def complete():
    arr1=[1,3,5,7]
    arr2=[2,4,6,8]
    i = 0
    j = 0
    result = []
    while i < len(arr1) and j < len(arr2):
            if arr1[i]<arr2[j]:
                result.append(arr1[i])
                i=i+1
            else:
                result.append(arr2[j])
                j=j+1

    while i < len(arr1):
        result.append(arr1[i])
        i = i + 1

    while j < len(arr2):
        result.append(arr2[j])
        j = j + 1

    return result
    
print("The Sorted Array is:",complete())

```
