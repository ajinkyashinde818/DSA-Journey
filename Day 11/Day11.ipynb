# 🧩 Day 11 — Remove Duplicates from a Sorted Array

## You are given a sorted array. Remove the duplicate values so that every value appears only once.

Example 1
arr = [1, 1, 2, 2, 3, 4, 4]

After removing duplicates:

[1, 2, 3, 4]

```python
def duplicate(array):
    result=[]
    for i in range(len(array)):
        if array[i] not in result:
            result.append(array[i])

    return result    
array=[1,1,2,2,3,4,4]
print("The Unique list is:",duplicate(array))
```
