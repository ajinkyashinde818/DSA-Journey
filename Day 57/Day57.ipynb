# Day 57: Find All Numbers Disappeared in an Array

## Problem Statement

Given an array `nums` of length `n`, where `nums[i]` is an integer in the range `[1, n]`, some elements appear once and others appear twice.

Return a list of all integers in the range `[1, n]` that do not appear in `nums`.

You must solve this problem with **O(n)** time complexity.

## Example 1

**Input:**
```python
nums = [4, 3, 2, 7, 8, 2, 3, 1]
```

**Output:**
```python
[5, 6]
```

**Explanation:** The numbers `5` and `6` do not appear in the array.

## Example 2

**Input:**
```python
nums = [1, 1]
```

**Output:**
```python
[2]
```

## Constraints

- `n == len(nums)`
- `1 <= n <= 10^5`
- `1 <= nums[i] <= n`
- Each element in `nums` appears once or twice.

## Hints

## Practice Goals

**Target Time Complexity:** `O(n)`

**Challenge:** Try to solve it without looking at the solution. Write your own logic, test the examples, and explain your approach before optimizing it.

````python
def disappered(nums):
    count=1
    temp=set(nums)
    result=[]
    
    for count in range(1,len(nums)+1):
        if count not in temp:
            result.append(count)

    return result

nums = [4, 3, 2, 7, 8, 2, 3, 1]
print("The Result is:",disappered(nums))
````
