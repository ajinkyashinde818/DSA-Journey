### 🔥 Day 42 — Reverse Integer

Given a signed 32-bit integer `x`, reverse its digits and return the reversed integer.

If reversing the integer causes it to go outside the signed 32-bit integer range `[-2³¹, 2³¹ - 1]`, return `0`.

**Examples:**

**Example 1:**
```python
Input: x = 123
Output: 321
```

**Example 2:**
```python
Input: x = -123
Output: -321
```

**Example 3:**
```python
Input: x = 120
Output: 21
```

**Example 4:**
```python
Input: x = 1534236469
Output: 0
```

**Constraints:**
- `-2³¹ <= x <= 2³¹ - 1`
- The environment does not allow storing 64-bit integers.

**Your Task:** Reverse the digits of the given integer while preserving its sign and checking for integer overflow.

````python
def reversed(num):
    rev=0
    while num>0:
        last=num%10
        if rev > 214748364 or (rev == 214748364 and num > 7):
            return 0
        rev=(rev*10)+last
        num=num//10

    
    return rev

num=1234567890
print("The 32-bit Reversed Number is:",reversed(num))
````
