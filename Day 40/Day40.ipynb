Product of Array Except Self

Given an integer array, return a new array where each element is the product of every other element, excluding itself.

Input:

[1, 2, 3, 4]

Output:

[24, 12, 8, 6]


Explanation:
- Index 0: \(2 \times 3 \times 4 = 24\)

- Index 1: \(1 \times 3 \times 4 = 12\)

- Index 2: \(1 \times 2 \times 4 = 8\)

- Index 3: \(1 \times 2 \times 3 = 6\)

```python
def product(arr):
    another=[]
    for i in range(len(arr)):
        num=1
        for j in range(0,len(arr)):
            if i!=j:
                num=num*arr[j]
        another.append(num)
    return another

arr=[1,2,3,4]
print("The Product of Array is:",product(arr))
```
