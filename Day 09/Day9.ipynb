
# 🧩 Day 9 — Find the Majority Element

## Given an array, find the element that appears more than half of the time.

Example
arr = [2, 2, 1, 2, 3, 2, 2]

Array length = 7

Half = 7 / 2 = 3.5

2 appears 5 times → 5 > 3.5 ✅

```python
def half_repeated(arr):
    length=len(arr)
    half=length/2

    for i in range(len(arr)):
        count=0
        for j in range(i,len(arr)):
            if arr[i]==arr[j]:
                count=count+1
        if count>half:
            return arr[i]
        
arr=[3, 3, 4, 3, 2]
print("The Majority Element is:",half_repeated(arr))
```
