## 🔥 Day 29 — Find the First Element That Appears Twice

Given:

arr = [2, 1, 3, 5, 3, 2]

Find the first element whose second occurrence appears earliest.

Here:

2 → appears at index 0 and 5

3 → appears at index 2 and 4

The second occurrence of 3 comes first, so:

Answer = 3

```python
def twice(array):
    found=[]
    

    for i in range(len(array)):
        for j in range(i+1,len(array)):
            if array[i]==array[j]:
                found.append(j)
    
    smallest=found[0]
    for i in range(1, len(found)):
        if found[i] < smallest:
            smallest = found[i]
        
    
    return array[smallest]

array=[2,1,3,5,3,2]
print("The Position is:",twice(array))
```
