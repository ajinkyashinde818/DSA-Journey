# 🧩 Day 13 — Find the First Repeating Element

## Given an array, find the first element that appears more than once.

Example 1
arr = [5, 3, 4, 3, 5, 2]

5 → appears again
3 → appears again

Answer → 3

```python
def first_repeating(arr):
    seen = set()

    for i in range(len(arr)):
        if arr[i] in seen:
            return arr[i]
        else:
            seen.add(arr[i])

    return None


arr = [5, 3, 4, 3, 5, 2]

print("First Repeating Element:", first_repeating(arr))
```
