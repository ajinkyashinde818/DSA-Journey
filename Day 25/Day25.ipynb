## 🔥 Day 25 — Find the Longest Increasing Subarray

Given:

arr = [1, 2, 3, 2, 4, 5, 6, 1]

Find the length of the longest continuous subarray where each element is greater than the previous element.

For example:

[1, 2, 3]       → length 3
[2, 4, 5, 6]    → length 4 ✅

So the answer is:

4

```python
def longest_array(arr):
    currentlength=1
    maxlength=1
    
    for i in range(1,len(arr)):
        if arr[i]>arr[i-1]:
            currentlength+=1
        else:
            currentlength=1

        if currentlength>maxlength:
            maxlength=currentlength

    return maxlength

arr=[1,2,3,2,4,5,6,1]
print("The Longest Subarray size is:",longest_array(arr))
```
