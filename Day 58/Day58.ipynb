# Day 57: Longest Subarray of 1's After Deleting One Element

## Problem Statement

Given a binary array `nums`, you must delete exactly one element from the array.

Return the length of the longest non-empty subarray containing only `1`s after deleting exactly one element.

A subarray consists of consecutive elements from the original array.

## Example 1

**Input:**
```python
nums = [1, 1, 0, 1]
```

**Output:**
```python
3
```

**Explanation:** Delete the zero to obtain `[1, 1, 1]`. The longest sequence contains 3 ones.

## Example 2

**Input:**
```python
nums = [0, 1, 1, 1, 0, 1, 1, 0, 1]
```

**Output:**
```python
5
```

**Explanation:** After deleting one element, the longest continuous sequence of ones has length 5.

## Example 3

**Input:**
```python
nums = [1, 1, 1]
```

**Output:**
```python
2
```

**Explanation:** We must delete exactly one element, even when every element is `1`.

## Constraints

- `1 <= nums.length <= 10^5`
- `nums[i]` is either `0` or `1`.

## Hints

1. Use two pointers: `left` and `right`.
2. Maintain a `zero_count` variable to count zeros in the current window.
3. Allow at most one zero in the window.
4. If `zero_count` becomes greater than one, move the `left` pointer forward until the window contains at most one zero.
5. Calculate the current window length and subtract one because exactly one element must be deleted.
6. Maintain `max_length` to store the maximum valid length.

## Approach

Use the **Sliding Window Technique** to maintain a continuous section of the array containing at most one zero.

### Variables

- `left`: Starting index of the current window.
- `zero_count`: Number of zeros in the current window.
- `max_length`: Maximum valid subarray length found so far.

## Complexity Target

- **Time Complexity:** `O(n)`
- **Extra Space Complexity:** `O(1)`

## Practice Goals

**Challenge:** Solve the problem using your own logic, test all examples, and explain why moving the left pointer is necessary when the window contains more than one zero.

````python
def longest_subarray(nums):
    left=0
    zero_count = 0
    max_length=0
    current_length=len(nums)

    for i in range(len(nums)):
        if nums[i]==0:
            zero_count+=1
            

        while zero_count>1:
            if nums[left]==0:
                zero_count-=1
            left=left+1

        current_length = i - left + 1
        valid_length = current_length - 1

        if valid_length > max_length:
            max_length = valid_length

    return max_length

nums = [0,1,1,1,0,1,1,0,1]
print("The Result is:",longest_subarray(nums))
````
