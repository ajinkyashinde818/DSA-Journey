### 🔥 Day 44 — String to Integer (atoi)

Given a string `s`, convert it into a 32-bit signed integer without using any built-in string-to-number conversion functions.

**Follow these steps:**

1. **Whitespace:** Ignore all leading spaces.
2. **Sign:** Check for `+` or `-`. If neither is present, assume the number is positive.
3. **Conversion:** Read consecutive digits and convert them into an integer manually. Stop at the first non-digit character. If no digits are found, return `0`.
4. **Overflow:** If the result exceeds the signed 32-bit integer range `[-2³¹, 2³¹ - 1]`, clamp it to the nearest limit.

**Examples:**

**Example 1:**
```python
Input: s = "42"
Output: 42
```

**Example 2:**
```python
Input: s = " -042"
Output: -42
```

**Example 3:**
```python
Input: s = "1337c0d3"
Output: 1337
```

**Example 4:**
```python
Input: s = "0-1"
Output: 0
```

**Example 5:**
```python
Input: s = "words and 987"
Output: 0
```

**Example 6:**
```python
Input: s = "91283472332"
Output: 2147483647
```

**Example 7:**
```python
Input: s = "-91283472332"
Output: -2147483648
```

**Constraints:**
- `0 <= len(s) <= 200`
- `s` contains English letters, digits, spaces, `+`, `-`, and `.`.
- Do not use built-in conversion functions such as `int()`, `atoi()`, or `parseInt()`.

**Your Task:** Implement the conversion manually using string traversal, digit extraction, sign handling, and overflow checking.

````python
def myAtoi(s):
    i = 0
    n = len(s)
    sign = 1
    result = 0

    INT_MAX = 2147483647
    INT_MIN = -2147483648

    while i < n and s[i] == ' ':
        i += 1

    if i < n and s[i] == '-':
        sign = -1
        i += 1
    elif i < n and s[i] == '+':
        sign = 1
        i += 1

    while i < n and '0' <= s[i] <= '9':
        digit = ord(s[i]) - ord('0')

        limit = 2147483648 if sign == -1 else 2147483647

        if result > limit // 10 or (
            result == limit // 10 and digit > limit % 10
        ):
            return INT_MIN if sign == -1 else INT_MAX

        result = result * 10 + digit
        i += 1

    return result * sign

print(myAtoi("42"))                 # 2
print(myAtoi(" -042"))              # -42
print(myAtoi("1337c0d3"))            # 1337
print(myAtoi("0-1"))                # 0
print(myAtoi("words and 987"))       # 0
print(myAtoi("91283472332"))         # 2147483647
print(myAtoi("-91283472332"))        # -2147483648

````
