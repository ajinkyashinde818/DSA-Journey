## 🔥Evaluate Infix to Postfix Expression
Given a Infix expression, evaluate its final result into a Postfix Expression using a Stack.

The Tokens are: ['A', '+', 'B', '*', 'C']

The Output Postfix: A B C * +

Result: A B C * +

```python

MAX = 100
stack = []


def push(data):
    if len(stack) == MAX:
        print("The stack is full")
    else:
        stack.append(data)


def pop():
    if len(stack) == 0:
        return None
    return stack.pop()


def precedence(operator):
    if operator == '^':
        return 3
    elif operator in ['*', '/', '%']:
        return 2
    elif operator in ['+', '-']:
        return 1
    return 0


def evaluate_postfix(expression):
    global stack
    stack = []

    tokens = expression.split()
    print("The Tokens are:", tokens)

    output = []
    operators = {'+', '-', '*', '/', '%', '^'}

    for token in tokens:

        # Step 1: Print operands as they arrive
        if token.isalnum():
            output.append(token)

        # Step 2: Push left parenthesis
        elif token == '(':
            push(token)

        # Step 3: Pop until left parenthesis
        elif token == ')':
            while stack and stack[-1] != '(':
                output.append(pop())

            if not stack:
                print("Error: Mismatched parentheses")
                return None

            pop()  # Discard '('

        # Step 4: Process operators
        elif token in operators:

            while stack and stack[-1] != '(':

                top = stack[-1]

                # Higher precedence on stack
                if precedence(top) > precedence(token):
                    output.append(pop())

                # Equal precedence: check associativity
                elif precedence(top) == precedence(token):
                    if token == '^':
                        break
                    output.append(pop())

                # Incoming operator has higher precedence
                else:
                    break

            push(token)

        else:
            print("Invalid token:", token)
            return None

    # Step 5: Pop all remaining operators
    while stack:
        if stack[-1] == '(':
            print("Error: Mismatched parentheses")
            return None

        output.append(pop())

    print("The Output Postfix:", " ".join(output))
    return " ".join(output)


if __name__ == "__main__":
    expression = input("Enter Infix Expression: ")
    answer = evaluate_postfix(expression)
    print("Result:", answer)
```
