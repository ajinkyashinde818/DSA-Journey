### 🔥 Day 47 — Letter Combinations of a Phone Number

Given a string `digits` containing digits from `2` to `9`, return all possible letter combinations that the number could represent using a telephone keypad.

Each digit maps to a group of letters. Return the combinations in any order.

**Telephone Keypad Mapping:**

- `2` → `"abc"`
- `3` → `"def"`
- `4` → `"ghi"`
- `5` → `"jkl"`
- `6` → `"mno"`
- `7` → `"pqrs"`
- `8` → `"tuv"`
- `9` → `"wxyz"`

**Examples:**

**Example 1:**
```python
Input: digits = "23"
Output: ["ad", "ae", "af", "bd", "be", "bf", "cd", "ce", "cf"]
```

**Explanation:** Each letter from digit `2` combines with each letter from digit `3`, producing 9 combinations.

**Example 2:**
```python
Input: digits = "2"
Output: ["a", "b", "c"]
```

**Example 3:**
```python
Input: digits = "79"
Output: ["pw", "px", "py", "pz", "qw", "qx", "qy", "qz", "rw", "rx", "ry", "rz", "sw", "sx", "sy", "sz"]
```

**Constraints:**
- `1 <= len(digits) <= 4`
- Every digit is between `2` and `9`.

**Your Task:** Generate all possible letter combinations represented by the given digit string.

````python
def combination(num):
    mapping = {
    '2': 'abc',
    '3': 'def',
    '4': 'ghi',
    '5': 'jkl',
    '6': 'mno',
    '7': 'pqrs',
    '8': 'tuv',
    '9': 'wxyz'
}   

    result=[""]
    for i in range(len(num)):
        letter=mapping[num[i]]
        resulttemp = []

        for combination in result:
            for j in letter:
                resulttemp.append(combination+j)

        result=resulttemp
    return result

num="23"
print("The Combination is:",combination(num))
````
