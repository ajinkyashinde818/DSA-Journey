### 🔥 Day 27 — Find the Equilibrium Index

Given an array, find an index where:

Sum of elements on the left = Sum of elements on the right

Example:

arr = [1, 3, 5, 2, 2]

Check index 2:

Left:  1 + 3 = 4

Right: 2 + 2 = 4

So:

Equilibrium Index = 2

```python
def equilibriam(array):
    total_sum=sum(array)
    leftsum=0
    rightsum=0
    for i in range(len(array)):
        rightsum = total_sum - leftsum - array[i]

        if leftsum==rightsum:
            return i
        
        leftsum=leftsum+array[i]
array=[1,3,5,2,2]
print("The Equilibriam Index is:",equilibriam(array))
```
