## 🧩 Day 42 — Median of Two Sorted Arrays

The median is the middle value after combining two sorted arrays.

Example 1:

nums1 = [1, 3]

nums2 = [2]

Combined: [1, 2, 3]

Median = 2


Example 2:

nums1 = [1, 2]

nums2 = [3, 4]

Combined: [1, 2, 3, 4]

Median = (2 + 3) / 2 = 2.5

```python
def sorted(arr):
    if len(arr)<=1:
        return arr
    pivot=arr[0]
    left=[]
    right=[]

    for i in range(1,len(arr)):
        if arr[i]<pivot:
            left.append(arr[i])
        else:
            right.append(arr[i])

    return sorted(left)+[pivot]+sorted(right)
arr1 = [3, 7, 12, 18, 24, 31, 39, 45, 52, 61,
        68, 74, 83, 91, 105, 117, 126, 138, 149, 160]
arr2 = [1, 5, 10, 15, 21, 28, 35, 42, 49, 57,
        65, 72, 80, 88, 99, 110, 122, 135, 147, 175]
arr=arr1+arr2
array = sorted(arr)
n = len(array)
if n % 2 == 1:
    median = array[n // 2]
else:
    median = (array[n // 2 - 1] + array[n // 2]) / 2
print("Median is:", median)
```
