## 🔥 Day 23 — Find the Longest Subarray with Sum K

Given an array:

arr = [1, 2, 1, 1, 1, 3, 2]
K = 5

Find the length of the longest subarray whose sum is exactly K.

For example:

[1, 2, 1, 1] → sum = 5

```python
def subarray(arr):
    target=5
    count=0
    for i in range(len(arr)):
        currentsum=0

        for j in range(i,len(arr)):
            currentsum=currentsum+arr[j]

            if currentsum == target:
                length = j - i + 1

                if length > count:
                    count = length
    
    return count

arr=[1,2,1,1,1,3,2]
print("The Count is:",subarray(arr))
```
