### 🎯 Problem: Reverse an Array

Given:

[1, 2, 3, 4, 5]

You need to reverse it:

[5, 4, 3, 2, 1]

```python
def reverse(arr):

    if len(arr) <= 1:
        return arr

    left = 0
    right = len(arr) - 1

    while left < right:

        temp = arr[left]
        arr[left] = arr[right]
        arr[right] = temp

        left = left + 1
        right = right - 1

    return arr


arr = [1, 2, 3, 4, 5]

print("Reverse:", reverse(arr))
```
