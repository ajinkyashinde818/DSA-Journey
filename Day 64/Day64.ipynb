## Problem
Implement Preorder Traversal of a Binary Tree using **recursion**.

**Order:** Root → Left → Right

## Example

```text
        1
       / \
      2   3
     / \
    4   5
```

**Output:**
```text
1 2 4 5 3
```


**Time:** O(n)  
**Space:** O(h), where `h` is the tree height.

**Goal:** Understand how recursion follows the Root → Left → Right order.

````python
class Node:
    def __init__(self,data):
        self.data=data
        self.left=None
        self.right=None
        
def preorder(tree):
    current = tree

    if tree is None:
        return 


    print(tree.data, end=" ")
    preorder(current.left)
    preorder(current.right)


            
tree = Node(1)
tree.left = Node(2)
tree.right = Node(3)
tree.left.left = Node(4)
tree.left.right = Node(5)

print("The Preorder Traversal is:",end=" ")
preorder(tree)
````
