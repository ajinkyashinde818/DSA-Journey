# 🎯 Day 5 Challenge: Move All Zeros to the End

## Given an array, move all 0s to the end while keeping the order of the other elements the same.

Example:
[1, 0, 2, 0, 3, 4]

Expected result:

[1, 2, 3, 4, 0, 0]

```python
def move_zeros(arr):
    numbers=[]
    zeros=[]

    for i in range(len(arr)):
        if arr[i] !=0:
            numbers.append(arr[i])
        else:
            zeros.append(arr[i])
    
    final=numbers+zeros
    return final

arr=[1, 0, 2, 0, 3, 4]
print("The sorted Numbers:",move_zeros(arr))
```
