# 🎯 Day 4 Challenge: Find the Missing Number

## You are given an array containing numbers from 1 to N, but one number is missing.

Example:
[1, 2, 3, 5, 6]

Which number is missing?

👉 4

Another example:

[1, 2, 4, 5]

👉 3

```python
def missing(arr):

    n=6
    expectedsum=n*(n+1)/2
    actualsum=0

    for i in range(len(arr)):
        
        actualsum=actualsum+arr[i]
       
    missing=expectedsum-actualsum

    return missing
arr = [1, 2, 3, 4, 6]    
print("The Missing Element is:",missing(arr))   
```
