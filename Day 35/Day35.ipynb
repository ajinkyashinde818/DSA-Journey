## 🧩 Day 35 — Valid Parentheses
Given a string containing only:

( ) { } [ ]

determine whether the brackets are valid/balanced.

Examples

Input:  "({[]})"

Output: true

Input:  "([)]"

Output: false

```python
MAX = 100
stack = []


def push(data):
    if len(stack) == MAX:
        print("The stack is full")
    else:
        stack.append(data)


def pop():
    if len(stack) == 0:
        return None
    return stack.pop()


def is_valid(expression):
    for char in expression:

        if char == '(' or char == '{' or char == '[':
            push(char)

        elif char == ')' or char == '}' or char == ']':

            if len(stack) == 0:
                return False

            top = stack[-1]

            if (char == ')' and top == '(') or \
               (char == '}' and top == '{') or \
               (char == ']' and top == '['):

                pop()

            else:
                return False

    return len(stack) == 0


if __name__ == "__main__":

    expression = input("Enter brackets: ")

    if is_valid(expression):
        print("Valid Parentheses")
    else:
        print("Invalid Parentheses")
```
