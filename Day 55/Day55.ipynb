
# Day 55: Container With Most Water

**Problem Statement:**

You are given an integer array `height` of length `n`. There are `n` vertical lines where the endpoints of the `i-th` line are `(i, 0)` and `(i, height[i])`.

Find two lines that, together with the x-axis, form a container that holds the maximum amount of water.

Return the maximum amount of water the container can store.

### Example 1
**Input:**
```python
height = [1, 8, 6, 2, 5, 4, 8, 3, 7]
```

**Output:**
```text
49
```

### Example 2
**Input:**
```python
height = [1, 1]
```

**Output:**
```text
1
```

### Constraints
- `2 <= height.length <= 10^5`
- `0 <= height[i] <= 10^4`

### Topics
- Array
- Two Pointers
- Greedy

### Goal
Solve the problem in Python with **O(n) time complexity** and **O(1) extra space**.


````python
def container(height):
    left=0
    right=len(height)-1
    width=right-left
    max_Area=0
   

    while left<right:
        width=right-left
        min_height=min(height[left],height[right])
        area=width*min_height
        if area>max_Area:
            max_Area=area
        if height[left] < height[right]:
            left=left+1
        else:
            right=right-1
    
    return max_Area    

height = [1, 8, 6, 2, 5, 4, 8, 3, 7]
print("The Container Size is:",container(height))
````
