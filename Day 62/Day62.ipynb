# Day 62: Non-Recursive Preorder Traversal of a Binary Tree


## Problem Statement

Implement a **non-recursive (iterative) Preorder Traversal** of a Binary Tree using a Stack.

In Preorder Traversal, visit the nodes in the following order:

**Root → Left → Right**

Your implementation must use an explicit stack instead of recursion to traverse the tree.

## Example

Consider the following binary tree:

```text
        1
       / \
      2   3
     / \
    4   5
```

**Output:**

```text
1 2 4 5 3
```

## Approach: Using a Stack

1. If the root is `NULL`, return.
2. Create an empty stack.
3. Push the root node onto the stack.
4. While the stack is not empty:
   - Pop the top node.
   - Print its data.
   - Push its right child if it exists.
   - Push its left child if it exists.
5. Continue until the stack becomes empty.

## Important Concept

A stack follows **LIFO (Last In, First Out)**.

The right child is pushed first, and the left child is pushed second. This ensures that the left child is processed before the right child.

## Complexity Analysis

- **Time Complexity:** O(n), where n is the total number of nodes.
- **Space Complexity:** O(n) in the worst case for the stack.

````python
class Node:
    def __init__(self,data):
        self.data=data
        self.left=None
        self.right=None
        
def inorder(tree):
    stack = []
    current = tree

    while stack or current is not None:
        if current is not None:
            stack.append(current)
            current = current.left
        else:
            current = stack.pop()
            print(current.data, end=" ")
            current = current.right
            
tree = Node(1)
tree.left = Node(2)
tree.right = Node(3)
tree.left.left = Node(4)
tree.left.right = Node(5)

print("The Inorder Traversal is:", end=" ")
inorder(tree)
````
