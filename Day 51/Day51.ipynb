🔥 Day 51 - Search a 2D Matrix.

Problem statement: You are given an \(m \times n\) matrix of integers with these properties:
 Each row is sorted in ascending order.
 The first integer of each row is greater than the last integer of the previous row.
Given an integer target, return True if the target exists in the matrix; otherwise, return False.

Example 1

matrix =[  [1,  3,  5,  7],    
            [10, 11, 16, 20],    
            [23, 30, 34, 60]
        ]
    
target = 3

Output:

True


Example 2
matrix =[  [1,  3,  5,  7],    
            [10, 11, 16, 20],    
            [23, 30, 34, 60]
        ]

target=13

Output:
False

```python
def matrix_target(matrix,target):
    for i in range(len(matrix)):
        for j in range(len(matrix[i])):
            if target==matrix[i][j]:
                return True

    return False

matrix =[  
    [1,  3,  5,  7],    
    [10, 11, 16, 20],    
    [23, 30, 34, 60]
]
target=3

print("The Result is:",matrix_target(matrix,target))
```
