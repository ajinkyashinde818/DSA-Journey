# Day 66: Maximum Depth of Binary Tree

**Date:** 07 November 2026  
**Topic:** Binary Tree / Recursion

## Problem
Given the root of a binary tree, find its **maximum depth**.

The maximum depth is the number of nodes along the longest path from the root node down to the farthest leaf node.

## Example

```text
        1
       / \
      2   3
     /
    4
```

**Output:**
```text
3
```

Because the longest path is:

`1 → 2 → 4`

**Time:** O(n)  
**Space:** O(h)

**Goal:** Understand how recursion can calculate information from the bottom of the tree back toward the root.
