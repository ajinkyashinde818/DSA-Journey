## 🔥 Day 26 — Find the First Missing Positive

Given:

arr = [3, 4, -1, 1]

Find the smallest positive number that is missing.

Positive numbers start from:

1, 2, 3, 4, 5, ...

In the array:

1 ✅

2 ❌

3 ✅

4 ✅

So the answer is:

2

```python
def missing(arr):
    missing=1
    found=None

    for i in range(len(arr)):
        found=False
        for j in range(len(arr)):
            if arr[j]==missing:
                found=True
                break
            
        if found==False:
            return missing
        else:
            missing=missing+1

    return missing
        

arr=[3,4,-1,1]
print("The Missing Element is:",missing(arr))
```
