# 🧩 Day 14 — Move All Negative Numbers to One Side

## Given an array containing positive and negative numbers, rearrange it so that all negative numbers #come to one side and positive numbers to the other side.

Example
arr = [1, -2, 3, -4, 5, -6]

One valid result:

[-2, -4, -6, 1, 3, 5]

The order does not matter.

Another valid result could be:

[-6, -2, -4, 5, 3, 1]

```python
def negative_repeat(arr):
    left=0
    right=len(arr)-1

    while left<right:

        if arr[left]<0:
            left=left+1

        if arr[right]>=0:
            right=right-1
        if left<right:

            arr[left],arr[right]=arr[right],arr[left]
            left = left + 1
            right = right - 1
    return arr

arr=[1,-2,3,-4,5,-6]
print("A Sorted Array is:",negative_repeat(arr))
```
