# 🧩 Day 16 — Find the Intersection of Two Arrays

## Given two arrays, find the elements that are present in both arrays.

Example 1

A = [1, 2, 3, 4, 5]

B = [3, 4, 5, 6, 7]

Answer = [3, 4, 5]

Example 2

A = [1, 2, 2, 3]

B = [2, 2, 4]

Answer = [2]

```python
def intersection():
    arr1=[1,2,3,4,5]
    arr2=[3,4,5,6,7]
    res=[]

    for i in range(len(arr1)):
        for j in range(len(arr2)):

            if arr1[i]==arr2[j] and arr1[i] not in res:
                res.append(arr1[i])

    return res
print("The Intersect Numbers:",intersection())
```
