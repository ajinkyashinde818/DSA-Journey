
# Day 60: Minimum Size Subarray Sum

## Problem Statement
Given an array of positive integers `nums` and a positive integer `target`, find the minimum length of a contiguous subarray whose sum is greater than or equal to `target`.

If no such subarray exists, return `0`.

## Examples

**Example 1**
- Input: `target = 7, nums = [2,3,1,2,4,3]`
- Output: `2`
- Explanation: The subarray `[4,3]` has a sum of `7` and length `2`.

**Example 2**
- Input: `target = 4, nums = [1,4,4]`
- Output: `1`
- Explanation: The subarray `[4]` has a sum of `4`.

**Example 3**
- Input: `target = 11, nums = [1,1,1,1,1,1,1,1]`
- Output: `0`
- Explanation: No subarray has a sum greater than or equal to `11`.

## Approach: Sliding Window

1. Initialize two pointers, `left` and `right`, at the beginning of the array.
2. Maintain a running sum of the current window.
3. Move `right` forward and add each element to the running sum.
4. Whenever the sum becomes greater than or equal to `target`, update the minimum window length.
5. Move `left` forward and subtract elements to make the window smaller.
6. Repeat until `right` reaches the end of the array.
7. Return the minimum length, or `0` if no valid window exists.

## Complexity Analysis

- **Time Complexity:** O(n), because each element is visited at most twice.
- **Space Complexity:** O(1), because only a few variables are used.


## Practice Goal
Solve the problem without using nested loops and aim for an O(n) solution.


```python
def minimum_size(nums,target):
    left=0
    right=0
    total=0
    min_length = float('inf')

    
    for right in range(len(nums)):
        total+=nums[right]

        
        while total>=target:
            current_length = right - left + 1
            if current_length < min_length:
                min_length = current_length
            total -= nums[left]
            left += 1

    if min_length == float('inf'):
        return 0

    return min_length

nums = [2, 3, 1, 2, 4, 3]
target = 7
print("The Result is:",minimum_size(nums,target))
```
