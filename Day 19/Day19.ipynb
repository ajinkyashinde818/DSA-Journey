# Day 19 — Find the Missing Number Using XOR

## Given an array containing numbers from 1 to n, with one number missing, find the missing number.

Example
Array = [1, 2, 4, 5, 6]

Numbers should be:

1, 2, 3, 4, 5, 6

So:

Answer = 3

```python
def missing(arr):
    result=0
    for i in arr:
        result=result^i

    for j in range(1,len(arr)+2):
         result = result ^ j


    return result


arr=[1,2,4,5,6]
print("The Number is:",missing(arr))
```
