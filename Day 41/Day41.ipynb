## 🧩 Day 41 - The String is a Palindrome or not
A palindrome is a string that reads the same forward and backward.
Examples:
- madam → Palindrome
- level → Palindrome
- hello → Not a palindrome

```python
MAX = 100
stack = []


def push(data):
    if len(stack) == MAX:
        print("The stack is full")
    else:
        stack.append(data)


def pop():
    if len(stack) == 0:
        return None
    return stack.pop()

def name(uinput):
    cleaned_input="".join(uinput.split()).lower()

    for char in cleaned_input:
        push(char)

    reversed=""
    for i in range(len(cleaned_input)):
        reversed=reversed+pop()
    
    if reversed==cleaned_input:
        print(cleaned_input,f":The String is Palindrome")
    else:
        print(cleaned_input,f":The String is not Palindrome")

if __name__ == "__main__":
    uinput = input("Enter a String:")
    name(uinput)
```
