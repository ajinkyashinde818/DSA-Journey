Find Minimum in Rotated Sorted Array
Medium
Topic: Binary Search

Goal: Find the smallest element in a rotated, sorted array in \(O(\log n)\) time.


Problem statement
You are given an array of distinct integers that was originally sorted in ascending order, but then rotated at an unknown pivot.

Your task is to find and return the minimum element in the array.
Example 1

arr = [3, 4, 5, 1, 2]

Output:1

Example 2

arr = [4, 5, 6, 7, 0, 1, 2]

Output:
0

```python
def rotated_minimum(arr):
    left=0
    right=len(arr)-1

    while left<right:
        mid = (left + right) // 2
        
        if arr[mid]>arr[right]:
            left=mid+1
        else:
            right=mid 
            
    return arr[left]

arr=[4, 5, 6, 7, 0, 1, 2]
print("The Minimum Element is:",rotated_minimum(arr))
```
