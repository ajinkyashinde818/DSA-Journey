# 🧩 Day 12 — Find the Element with the Highest Frequency

## Given an array, find the element that appears the most number of times.

Example 1
arr = [1, 2, 2, 3, 1, 2, 4]


1 → 2 times

2 → 3 times ✅

3 → 1 time

4 → 1 time

Answer → 2

```python
def multitime_repeated(arr):
    frequency = {}
    highfreq = 0
    answer_element = None

    for i in range(len(arr)):
        if arr[i] in frequency:
            frequency[arr[i]] = frequency[arr[i]] + 1
        else:
            frequency[arr[i]] = 1

        current_frequency = frequency[arr[i]]

        if current_frequency > highfreq:
            highfreq = current_frequency
            answer_element = arr[i]

    return answer_element


arr = [1, 2, 2, 3, 1, 2, 4]

print("Highest Frequency Element:", multitime_repeated(arr))
```
