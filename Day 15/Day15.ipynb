# 🧩 Day 15 — Find the Longest Consecutive Sequence

## Given an array of integers, find the length of the longest sequence of consecutive numbers.

Example 1
Input:
[100, 4, 200, 1, 3, 2]

Consecutive sequence:
1, 2, 3, 4

Answer:
4

```python
def consecutive_num(array):
    max_length=0
    for i in range(len(array)):
        current_length=1
        next_number=array[i]+1
        found=True

        while found:
            found=False
            for j in range(len(array)):
                if array[j]==next_number:
                    current_length=current_length+1
                    next_number=next_number+1
                    found=True
                    break

        if current_length>max_length:
            max_length=current_length

    return max_length 


array=[100, 4, 200, 1, 3, 2]
print("The length is:",consecutive_num(array))
```
