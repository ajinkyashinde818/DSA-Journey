🎯 Sort the give array using a Quick sort

pivot=arr[0]

arr=[11,2,4,13,57,65,47,1,50]

ans 

arr=[1,2,4,11,13,47,5057,65]

```python
def quick(arr):
    if len(arr)<=1:
        return arr
    left=[]
    right=[]
    pivot=arr[0]

    for i in range(1,len(arr)):
        if arr[i]<pivot:
            left.append(arr[i])
        else:
            right.append(arr[i])

    return quick(left)+[pivot]+quick(right)

arr=[11,2,4,13,57,65,47,1,50]
print("The Sorted Element is:",quick(arr))
    
```
