# 🎯 Day 7: Two Sum

## Given an array and a target value, find two elements whose sum equals the target.

Example:

arr = [2, 7, 11, 15]
target = 9

Expected:

2 + 7 = 9

So the answer could be:

[2, 7]

```python
def checking(arr):
    target=9

    for i in range(len(arr)):
        for j in range(i+1,len(arr)):
            sum =arr[i]+arr[j]

            if sum==target:
                return arr[i],arr[j]
            

arr=[2,7,11,15]
print("The correct Pair is:",checking(arr))    
```
