# Day 74: Validate Binary Search Tree

## Problem
Given the root of a binary tree, determine whether it is a **valid Binary Search Tree (BST)**.

A valid BST must follow:
- All values in the left subtree are smaller than the node.
- All values in the right subtree are greater than the node.

## Example

```text
        2
       / \
      1   3
```

**Output:**
```text
True
```

```text
        5
       / \
      1   4
         / \
        3   6
```

**Output:**
```text
False
```

## Approach
Use recursion and keep track of the valid range for each node.

1. Start with a range of `-∞` to `+∞`.
2. The current node must be inside this range.
3. For the left subtree, update the upper limit.
4. For the right subtree, update the lower limit.
5. Continue recursively for every node.

**Time:** O(n)  
**Space:** O(h)

**Goal:** Understand how recursion can maintain valid boundaries while checking a BST.

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def is_valid_bst(root, min_value, max_value):

    if root is None:
        return True

    if root.data <= min_value or root.data >= max_value:
        return False

    left_valid = is_valid_bst(
        root.left,
        min_value,
        root.data
    )

    right_valid = is_valid_bst(
        root.right,
        root.data,
        max_value
    )

    return left_valid and right_valid


# Create BST
tree = Node(2)
tree.left = Node(1)
tree.right = Node(3)

result = is_valid_bst(tree, float("-inf"), float("inf"))

print("Is Valid BST:", result)
````
