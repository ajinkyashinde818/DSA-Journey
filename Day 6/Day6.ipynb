# 🎯 Day 6 Challenge: Find Duplicate Elements

## Given an array, find the elements that appear more than once.

Example:
arr = [1, 2, 3, 2, 4, 5, 1]

Duplicates are:

1, 2

Another example:

arr = [10, 20, 30, 40]

Result:

No duplicates

```python
def duplicate(arr):
    duplicate=[]

    for i in range(len(arr)):
        for j in range(i+1,len(arr)):
            if arr[i]==arr[j]:
                if arr[i] not in duplicate:
                    duplicate.append(arr[i])
    
    return duplicate
                
arr = [1, 2, 3, 2, 4, 5, 1]
print("Duplicate Elements:",duplicate(arr))
```
