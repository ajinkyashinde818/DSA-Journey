# 🧩 Day 10 — Find the Second Largest Unique Element

## Given an array, find the second largest distinct element.

Example 1
arr = [10, 5, 20, 8, 20]

Largest = 20
Second largest = 10

Answer → 10

Notice 20 appears twice, but we count it as one distinct value.

```python
# Find a Second Largest

def second_largest(arr):

    if len(arr) < 2:
        return None

    largest_number = arr[0]
    second_number = None

    for i in range(1, len(arr)):
        if arr[i] > largest_number:
            second_number = largest_number
            largest_number = arr[i]
        elif arr[i] < largest_number:
            if second_number is None or arr[i] > second_number:
                second_number = arr[i]
    return second_number

arr = [10,5,20,8,20]
print("Second Largest Number:", second_largest(arr))
```
