Find the Longest Subarray with Equal Number of 0s and 1s

Example:

Input:  [0, 1, 0, 0, 1, 1, 0]

Output: 6

The longest subarray is:

[1, 0, 0, 1, 1, 0]

```python
def subarray(arr):
    maximum = 0
    length=0

    for i in range(len(arr)):
        zero_count=0
        one_count=0
        for j in range(i,len(arr)):
            if arr[j]==0:
                zero_count+=1
            else:
                one_count+=1
            
            if zero_count==one_count:
                length = j - i + 1
            if length > maximum:
                maximum = length

    return maximum

arr=[0,1,0,0,1,1,0]
print("The Lenth of Subarray is:",subarray(arr))
```
