# 🎯 Day 3 Task: Check if an Array is Sorted

## Problem:

## Given an array, determine whether it is sorted in ascending order.



Examples:
[1, 2, 3, 4, 5] → True ✅

[1, 5, 3, 4, 2] → False ❌

[10, 20, 30, 40] → True ✅

```python
def sorting(arr):

    for i in range(len(arr)-1):
        if arr[i]>arr[i+1]:
            return "Array is Not sorted"
        
    return "Array is Sorted"
    
    
arr=[1,2,3,6]
print(sorting(arr))
```
