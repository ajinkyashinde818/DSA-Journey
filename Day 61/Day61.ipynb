
# Day 61: Max Consecutive Ones III

## Problem Statement
Given a binary array `nums` and an integer `k`, you can flip at most `k` zeros into ones.

Find the maximum number of consecutive ones in the array after performing at most `k` flips.

## Examples

**Example 1**
- Input: `nums = [1,1,0,0,1,1,1,0]`, `k = 2`
- Output: `7`
- Explanation: Flip two zeros to obtain a continuous subarray of five ones.

**Example 2**
- Input: `nums = [1,0,1,1,0,1]`, `k = 1`
- Output: `4`
- Explanation: Flip one zero to get four consecutive ones.

## Approach: Variable-Size Sliding Window

1. Initialize two pointers, `left` and `right`, at the beginning of the array.
2. Maintain a count of zeros in the current window.
3. Move `right` forward to expand the window.
4. If `nums[right] == 0`, increase the zero count.
5. If the zero count becomes greater than `k`, shrink the window by moving `left` forward.
6. If the element at `left` is zero, decrease the zero count before moving `left`.
7. Track the maximum window length using `right - left + 1`.
8. Return the maximum length.

## Complexity Analysis

- **Time Complexity:** O(n), because each element is visited at most twice.
- **Space Complexity:** O(1), because only a few variables are used.


```python
def longest_ones(nums, k):
    left = 0
    zero_count = 0
    max_length = 0

    for right in range(len(nums)):
        if nums[right]==0:
            zero_count=zero_count+1

        while zero_count > k:
            if nums[left]==0:
                zero_count-=1
            left=left+1

        current_length = right - left + 1

        if current_length > max_length:
            max_length = current_length
            
    return max_length

nums=[1,0,1,1,0,1]
k = 1
print("The Result is:", longest_ones(nums, k))
```
