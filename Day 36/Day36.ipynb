## 🔥Evaluate Postfix Expression
Given a postfix expression, evaluate its final result using a Stack.

Example:

Input: 
 
23*54*+9-

Output: 17

Because:
2 3 *     → 6

5 4 *     → 20

6 + 20    → 26

26 9 -    → 17

```python
MAX = 100
stack = []


def push(data):
    if len(stack) == MAX:
        print("The stack is full")
    else:
        stack.append(data)


def pop():
    if len(stack) == 0:
        return None
    return stack.pop()

def evaluate_postfix(expression):

    for char in expression:
        if char.isdigit():
            push(int(char))

        elif char in ['+', '-', '*', '/']:
            first = pop()
            second = pop()

            if char == '+':
                result = second + first
            elif char == '-':
                result = second - first
            elif char == '*':
                result = second * first
            elif char == '/':
                if first == 0:
                    print("Error: Division by zero")
                    return None
                result = second / first
            push(result)

    return pop()


if __name__ == "__main__":
    expression = input("Enter Expression: ")
    answer = evaluate_postfix(expression)
    print("Result:", answer)
```
