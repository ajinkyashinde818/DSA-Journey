**Topic:** Binary Tree / Recursion / Postorder Traversal

## Problem
Implement Postorder Traversal of a Binary Tree using recursion.

**Order:** Left → Right → Root

## Example

```text
        1
       / \
      2   3
     / \
    4   5
```

**Output:**
```text
4 5 2 3 1
```

**Time:** O(n)  
**Space:** O(h)

````python
class Node:
    def __init__(self,data):
        self.data=data
        self.left=None
        self.right=None
        
def postorder(tree):
    if tree is None:
        return 

    postorder(tree.left)
    postorder(tree.right)
    print(tree.data,end=" ")


            
tree = Node(1)
tree.left = Node(2)
tree.right = Node(3)
tree.left.left = Node(4)
tree.left.right = Node(5)

print("The Postorder Traversal is:",end=" ")
postorder(tree)
````
