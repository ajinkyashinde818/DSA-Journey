# Day 67: Same Tree

## Problem
Given the roots of two binary trees `p` and `q`, determine whether the two trees are exactly the same.

Two trees are the same if they have:
- The same structure
- The same node values

## Example

```text
Tree 1:        1          Tree 2:        1
              / \                      / \
             2   3                    2   3
```

**Output:**
```text
True
```

**Time:** O(n)  
**Space:** O(h)

**Goal:** Understand how recursion can compare two binary trees node by node.

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def is_same_tree(p, q):
    if p is None and q is None:
        return True

    if p is None or q is None:
        return False

    if p.data !=q.data:
        return False

    left_same = is_same_tree(p.left,q.left)
    right_same = is_same_tree(p.right,q.right)

    return left_same and right_same


# Tree 1
p = Node(1)
p.left = Node(2)
p.right = Node(3)

# Tree 2
q = Node(1)
q.left = Node(2)
q.right = Node(3)

print(is_same_tree(p, q))
````
