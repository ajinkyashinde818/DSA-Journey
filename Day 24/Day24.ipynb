## 🔥 Day 24 — Find the Maximum Consecutive 1s

Given a binary array containing only 0 and 1:

arr = [1, 1, 0, 1, 1, 1, 0, 1]

Find the maximum number of consecutive 1s.

Expected output:

3

Because:

1, 1, 0, 1, 1, 1, 0, 1

        └─────┘
            3


```python
def consecutive(arr):
    current,maximum=0,0
    for i in range(len(arr)):
        if arr[i]==1:
            current+=1
        else:
            current=0

        if current>maximum:
            maximum=current

    return maximum

arr=[1,1,0,1,1,1,0,1]
print("The Pais is:",consecutive(arr))
```
