Find the Maximum Product of Two Elements in an Array

Given an array of integers, find the maximum product that can be obtained by multiplying any two different elements.

Example:

arr = [3, 5, 2, 9, 7]

Output = 63

Because:

9 × 7 = 63

```python
def maximumnum(arr):
    max=0
    product=0
    for i in range(len(arr)):
        for j in range(i+1,len(arr)):
            product=arr[i]*arr[j]
            if product>max:
                max=product

    return max

arr=[3,5,2,9,7]
print("The Product is:",maximumnum(arr))
```
