### 🔥 Day 28 — Find the Maximum Difference Between Two Elements

Given:

arr = [2, 3, 10, 6, 4, 8, 1]

Find the maximum value of arr[j] - arr[i] where j > i.

Example:

10 - 2 = 8

So:

Maximum Difference = 8

```python
def diff(arr):
    minimum=arr[0]
    maxDifference=0

    for i in range(1,len(arr)):
        difference=arr[i]-minimum
        
        if difference>maxDifference:
            maxDifference=difference
        if arr[i]<minimum:
            minimum=arr[i]

    return maxDifference

arr=[2,3,10,6,4,8,1]
print("The Difference is:",diff(arr))
```
