# Day 56: Product of Array Except Self

**Problem:** Given an integer array `nums`, return an array `answer` such that `answer[i]` is equal to the product of all elements of `nums` except `nums[i]`.

### Example 1
**Input:**
```python
nums = [1, 2, 3, 4]
```
**Output:**
```python
[24, 12, 8, 6]
```

### Example 2
**Input:**
```python
nums = [-1, 1, 0, -3, 3]
```
**Output:**
```python
[0, 0, 9, 0, 0]
```

### Constraints
- \(2 \leq nums.length \leq 10^5\)
- \(-30 \leq nums[i] \leq 30\)
- The product of any prefix or suffix of `nums` is guaranteed to fit in a 32-bit integer.
- Do not use division.
- Solve in **\(O(n)\) time complexity**.
- Aim for **\(O(1)\) extra space**, excluding the output array.

### Topic
**Array | Prefix and Suffix Products**

### Practice Goal
Solve the problem independently in Python. First understand prefix products (elements to the left) and suffix products (elements to the right), then combine them to calculate each answer.

````python
def product(nums):
    n=len(nums)
    left=[]
    right=[]
    result=[]
    running_left=1
    right = [0] * len(nums)
    for i in range(len(nums)):
        left.append(running_left)
        running_left=running_left*nums[i]

    running_right = 1
    for i in reversed(range(len(nums))):
        right[i] = running_right             
        running_right = running_right * nums[i]

    for i in range(n):
        result.append(left[i] * right[i])
        
    return result
    

nums=[1,2,3,4]
print("The Result is:",product(nums))
````
